```python
from pathlib import Path
import importlib.metadata
import os
import platform
import sys

PROJECT_ROOT = next(
    (
        path
        for path in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
        if (path / "PROJECT_PLAN.md").is_file()
    ),
    None,
)
if PROJECT_ROOT is None:
    raise FileNotFoundError("Could not locate PROJECT_PLAN.md from the current working directory")

required_directories = (
    "datasets",
    "processed",
    "src",
    "configs",
    "models",
    "runs",
    "reports",
)
missing_directories = [
    name for name in required_directories if not (PROJECT_ROOT / name).is_dir()
]
if missing_directories:
    raise FileNotFoundError(f"Missing project directories: {missing_directories}")

package_versions = {}
for package_name in ("ultralytics", "torch", "opencv-python", "numpy"):
    try:
        package_versions[package_name] = importlib.metadata.version(package_name)
    except importlib.metadata.PackageNotFoundError:
        package_versions[package_name] = "NOT INSTALLED"

checks = {
    "processed_phone_dataset": (PROJECT_ROOT / "processed" / "phone_detection").is_dir(),
    "processed_defect_dataset": (PROJECT_ROOT / "processed" / "defect_segmentation").is_dir(),
}
environment_ready = (
    not missing_directories
    and all(version != "NOT INSTALLED" for version in package_versions.values())
    and all(checks.values())
)

print(f"Python version: {platform.python_version()}")
print(f"Python executable: {sys.executable}")
print(f"Project root: {PROJECT_ROOT}")
print(f"ultralytics version: {package_versions['ultralytics']}")
print(f"torch version: {package_versions['torch']}")
print(f"OpenCV version: {package_versions['opencv-python']}")
print(f"numpy version: {package_versions['numpy']}")
print(f"CPU: {platform.processor() or 'Unknown'} ({os.cpu_count() or 'Unknown'} logical cores)")
print(f"Processed phone dataset exists: {checks['processed_phone_dataset']}")
print(f"Processed defect dataset exists: {checks['processed_defect_dataset']}")
print("ENVIRONMENT READY" if environment_ready else "ENVIRONMENT NOT READY")
```
